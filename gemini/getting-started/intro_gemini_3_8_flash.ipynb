## 🎥 Agentic Video Understanding

**Agentic Video** is a multimodal processing approach where Gemini actively navigates a video timeline to inspect only the specific frames, audio, or captions needed for a query, rather than passively ingesting the entire video upfront. No manual video splitting or frame sampling setup is required.

You can use it to reduce token costs and latency by up to 95% on long-form content while gaining higher precision for fast-action, localized visual searches, and multi-video analysis.

When sending a video, set the `media_processing` field to `agentic`. Static processing is enabled by default.

For more information, refer to [Agentic Video Understanding](https://docs.cloud.google.com/gemini-enterprise-agent-platform/models/capabilities/video-understanding#agentic-video-processing).

````
response = client.models.generate_content(
    model=MODEL_ID,
    contents=[
        types.Part(
            file_data=types.FileData(
                file_uri="https://www.youtube.com/watch?v=LzExSq9DU9w",
                mime_type="video/mp4",
            ),
            media_processing="agentic",  # Options: "agentic", "static"
        ),
        "What were the key revenue figures mentioned by the presenter, and at what timestamp do they appear?",
    ],
    config=types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(
            thinking_level=types.ThinkingLevel.MEDIUM,  # LOW, MEDIUM, HIGH
        ),
    ),
)

display(Markdown(response.text.replace("$", "\\$")))
print(f"Total Token Count: {response.usage_metadata.total_token_count}")
````

## 🧠 Reasoning & Thinking Control

Gemini 3.8 Flash features a customizable reasoning depth. The **default thinking level is `high`**.

Developers can adjust the `thinking_level` parameter inside `ThinkingConfig` to fine-tune this balance.

| Thinking Level | Description | Recommended Use Case |
| :--- | :--- | :--- |
| `LOW` | Lightweight logical checks with high execution speed and low token cost. | Moderate concepts comparison, editing. |
| `MEDIUM` (Default) | Highly balanced reasoning effort. | Standard problem solving, standard logical deduction. |
| `HIGH` (Dynamic) | Maximizes internal chain-of-thought loops before first-token output. | Advanced math, logic puzzles, complex code generation. |

> ⚠️ **NOTE:** You cannot combine `thinking_level` with the legacy `thinking_budget` in the same configuration block. Doing so returns a `400 Bad Request` API error.

### Example: High-Level Reasoning for Logic Puzzles

Configure `thinking_level` to `HIGH` to solve a classic math puzzle.

````
prompt = """A farmer has chickens and rabbits. There are 35 heads and 94 legs. How many chickens and how many rabbits does he have? Show your step-by-step logic."""

response = client.models.generate_content(
    model=MODEL_ID,
    contents=prompt,
    config=types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(thinking_level=types.ThinkingLevel.HIGH)
    ),
)

display(Markdown(response.text))
````

### Extracting Thought Summaries (`include_thoughts`)

When using reasoning capabilities, you can inspect the model's inner thoughts separately from the final answer by setting `include_thoughts=True`.

````
response = client.models.generate_content(
    model=MODEL_ID,
    contents="How many R's are in the word strawberry?",
    config=types.GenerateContentConfig(
        thinking_config=types.ThinkingConfig(
            include_thoughts=True,
            thinking_level=types.ThinkingLevel.LOW,
        )
    ),
)

for part in response.candidates[0].content.parts:
    if part.thought:
        display(Markdown(f"""## Thoughts:n{part.text}"""))
    else:
        display(Markdown(f"""## Answer:\n{part.text}"""))
````

## 🎆 Multimodal Controls & Media Resolution

*   **Global Configuration:** Set a general resolution limit for all media in the request.
*   **Per-Part Resolution (Recommended):** Set separate media resolutions per item (e.g., highly detailed schema image in `ULTRA_HIGH` and video in `LOW`).

## 🤖 Agentic Workflows & Advanced Tooling

Gemini 3.7 Flash excels at scaling complex, long-horizon tasks.

When a reasoning model calls an external tool (Function Calling), it pauses its internal thinking. The **Thought Signature** is an encrypted token returned by the model acting as a "save state".

To maintain complete context and prevent logical fragmentation across agent boundaries, **you must return this signature** in subsequent conversation rounds back to the model. This preserves the reasoning chain and prevents tool overuse.

*   **Automatic Function Calling (Recommended):** The SDK manages thought signatures behind the scenes.
*   **Manual Function Calling:** The developer must explicitly append the previous candidate's content block (containing the signature) when assembling history.

### Example A: Automatic Function Calling (Thought Signatures Handled Automatically)

The SDK resolves intermediate execution paths and automatically manages signatures in the background.

````
def get_weather(city: str):
    """Gets the weather in a city."""
    if "london" in city.lower():
        return "Rainy"
    if "new york" in city.lower():
        return "Sunny"
    return "Cloudy"


response = client.models.generate_content(
    model=MODEL_ID,
    contents="What's the weather in London and New York?",
    config=types.GenerateContentConfig(
        tools=[get_weather],
    ),
)

# The SDK handles the function calls and thought signatures, and returns the final text
display(Markdown("## Final Response"))
display(Markdown(response.text))

# Print function calling history
hist_turn = response.automatic_function_calling_history[1]
print("\nFunction Call 1:", hist_turn.parts[1].function_call.name)
````

### Example B: Manual Function Calling (Explicit Thought Preservation)

Here, we manually run the tool and pass the signature block back in the history array to maintain the model's reasoning chain.

````
# 1. Define your tool
get_weather_declaration = types.FunctionDeclaration(
    name="get_weather",
    description="Gets the current weather temperature for a given location.",
    parameters={
        "type": "object",
        "properties": {"location": {"type": "string"}},
        "required": ["location"],
    },
)
get_weather_tool = types.Tool(function_declarations=[get_weather_declaration])

# 2. Send a message that triggers the tool
prompt = "What's the weather like in London?"
response = client.models.generate_content(
    model=MODEL_ID,
    contents=prompt,
    config=types.GenerateContentConfig(
        tools=[get_weather_tool],
        thinking_config=types.ThinkingConfig(include_thoughts=True),
    ),
)

# 3. Handle the function call
function_call = response.function_calls[0]
location = function_call.args["location"]
print(f"Model wants to call: {function_call.name}")

# Execute your tool (e.g., call an API)
# (This is a mock response for the example)
print(f"Calling external tool for: {location}")
function_response_data = {
    "location": location,
    "temperature": "30C",
}

# 4. Send the tool's result back
# Append this turn's messages to history for a final response.
# The `content` object automatically attaches the required thought_signature behind the scenes.
history = [
    types.Content(role="user", parts=[types.Part(text=prompt)]),
    response.candidates[0].content,  # Signature preserved here
    types.Content(
        role="tool",
        parts=[
            types.Part.from_function_response(
                name=function_call.name,
                response=function_response_data,
            )
        ],
    ),
]

response_2 = client.models.generate_content(
    model=MODEL_ID,
    contents=history,
    config=types.GenerateContentConfig(
        tools=[get_weather_tool],
        thinking_config=types.ThinkingConfig(include_thoughts=True),
    ),
)

# 5. Get the final, natural-language answer
print(f"\nFinal model response: {response_2.text}")
````

### Streaming Function Call Arguments

Turn on partial tool argument streaming using `stream_function_call_arguments=True` to improve streaming performance when calling tools.

````
get_weather_declaration = types.FunctionDeclaration(
    name="get_weather",
    description="Gets the current weather temperature for a given location.",
    parameters={
        "type": "object",
        "properties": {"location": {"type": "string"}},
        "required": ["location"],
    },
)
get_weather_tool = types.Tool(function_declarations=[get_weather_declaration])


for chunk in client.models.generate_content_stream(
    model=MODEL_ID,
    contents="What's the weather in London and New York?",
    config=types.GenerateContentConfig(
        tools=[get_weather_tool],
        tool_config=types.ToolConfig(
            function_calling_config=types.FunctionCallingConfig(
                mode=types.FunctionCallingConfigMode.AUTO,
                stream_function_call_arguments=True,
            )
        ),
    ),
):
    if chunk.function_calls:
        function_call = chunk.function_calls[0]
        if function_call and function_call.name:
            print(f"{function_call.name}")
            print(f"will_continue={function_call.will_continue}")
````

### Multimodal Function Responses

Return structured multimodal file references directly inside function call tool responses.

````
# 1. Define the function tool
get_image_declaration = types.FunctionDeclaration(
    name="get_image",
    description="Retrieves the image file reference for a specific order item.",
    parameters={
        "type": "object",
        "properties": {
            "item_name": {
                "type": "string",
                "description": "The name or description of the item ordered (e.g., 'green shirt').",
            }
        },
        "required": ["item_name"],
    },
)
tool_config = types.Tool(function_declarations=[get_image_declaration])

# 2. Send a message that triggers the tool
prompt = "Show me the green shirt I ordered last month."
response_1 = client.models.generate_content(
    model=MODEL_ID,
    contents=[prompt],
    config=types.GenerateContentConfig(
        tools=[tool_config],
    ),
)

# 3. Handle the function call
function_call = response_1.function_calls[0]
requested_item = function_call.args["item_name"]
print(f"Model wants to call: {function_call.name}")

# Execute your tool (e.g., call an API)
# (This is a mock response for the example)
print(f"Calling external tool for: {requested_item}")

function_response_data = {
    "image_ref": {"$ref": "dress.jpg"},
}

function_response_multimodal_data = types.FunctionResponsePart(
    file_data=types.FunctionResponseFileData(
        mime_type="image/png",
        display_name="dress.jpg",
        file_uri="gs://cloud-samples-data/generative-ai/image/dress.jpg",
    )
)

# 4. Send the tool's result back
# Append this turn's messages to history for a final response.
history = [
    types.Content(role="user", parts=[types.Part(text=prompt)]),
    response_1.candidates[0].content,
    types.Content(
        role="tool",
        parts=[
            types.Part.from_function_response(
                name=function_call.name,
                response=function_response_data,
                parts=[function_response_multimodal_data],
            )
        ],
    ),
]

response_2 = client.models.generate_content(
    model=MODEL_ID,
    contents=history,
    config=types.GenerateContentConfig(
        tools=[tool_config], thinking_config=types.ThinkingConfig(include_thoughts=True)
    ),
)

display(Markdown(f"\nFinal model response:\n{response_2.text}"))
````

## 🚉 Platform Features & Model Configurations

````
import os
import re
import random
import hashlib
import gradio as gr
from google import genai
from google.genai import types

# ==============================================================================
# 1. KERNEL & CLIENT INITIALIZATION (GEMINI 3.8 FLASH)
# ==============================================================================
MODEL_ID = "gemini-3.8-flash"
PROJECT_ID = os.getenv("GOOGLE_CLOUD_PROJECT", "gen-lang-client-0550120310")
client = genai.Client(enterprise=True, project=PROJECT_ID, location="global")


MASTER_KERNEL_10_0 = """====================================================================================================
VO5 SUNO OMNI-PRODUCER • BUILD 10.0.0 (SINGULARITY V6 IP-LOCK EDITION)
Architect: Sovanndara Diep | Mantra: "Ink is Value. Sound is Emotion. Structure is Power."
====================================================================================================
1. BOX 1 HARD CEILING: Strictly <= 120 characters, starting with [24-bit 48kHz].
2. BOX 2 DICTION LOCK: Always begin lyrics with the 4-line 24-bit high-resolution enunciation header.
3. LYRIC ENGINE: Enforce strict syllable counting (Chhanda) and heavy semantic metaphorical abstraction.
4. OUTPUT FORMAT: Explicitly label sections with === TITLE ===, === BOX 1 ===, and === BOX 2 ===.
5. IP-MOD 43: All tracks are cryptographically registered intellectual property of Sovanndara Diep.
"""

LOCKED_LINE_HASHES = set()

def compute_sha256_hex(text: str) -> str:
    return hashlib.sha256(text.encode("utf-8")).hexdigest()

def build_diction_header(vocal_str: str = "") -> str:
    v_low = str(vocal_str).lower()
    if "ឆ្លើយឆ្លង" in v_low or "duet" in v_low:
        v_profile = "Consistent male and female duet voices"
    elif "រ៉េប" in v_low or "rap" in v_low:
        v_profile = "Consistent raspy rhythmic rap vocal"
    elif "និទាន" in v_low or "narration" in v_low:
        v_profile = "Consistent deep cinematic narration vocal"
    elif "ស្រី" in v_low or "female" in v_low:
        v_profile = "Consistent female voice"
    elif "ប្រុស" in v_low or "male" in v_low:
        v_profile = "Consistent male voice"
    else:
        v_profile = "Consistent female voice"

    return (
        f"[Vocal Mix: 24-bit high-resolution, dry close-mic, hyper-distinct syllables]\n"
        f"[{v_profile}, perfect Khmer diction, precise enunciation]\n"
        f"[no spoke, no speech, enunciate every syllable clearly]\n"
        f"[Breath Control: Natural human breathing, smooth phrasing, zero vocal clipping]"
    )

MOOD_LOGIC_MATRIX = {
    "ឈឺចាប់ស៊ីជម្រៅ": {
        "genre": "[24-bit 48kHz] Epic Angkor cinematic ballad, weeping Tro Sau, Khloy flute, Skor Thom, 68 BPM",
        "vocal": "ស្រី (ផ្អែមស្រទន់)",
        "chhanda": "បទព្រហ្មគីតិ (៥-៦ ព្យាង្គ)",
        "energy": "មហិមា វិសេសវិសាល",
        "dna": "Deep Melancholic Resonance & Weeping Tro Sau Tears",
        "inst": ["Weeping Tro Sau", "Airy Khloy Flute", "Deep Skor Thom"],
        "nouns": ["ប្រាសាទថ្មភក់", "ក្រាំងស្បែកប្រើស", "ទឹកភ្នែកមហានគរ", "ស្លឹកឈើចាស់"],
        "modifiers": ["ដ៏ស្ងាត់ជ្រងំ", "ដែលបន្សល់ទុក", "ដ៏ឈឺចាប់ស៊ីជម្រៅ", "ក្រោមតំណក់ភ្លៀង"]
    },
    "មោទនភាពអ្នកចម្បាំង": {
        "genre": "[24-bit 48kHz] Epic Angkor war march, thunderous Skor Thom, Sralai oboe, brass, 85 BPM",
        "vocal": "ប្រុស (មាំទាំ រណ្តំ)",
        "chhanda": "បទភុជង្គលីលា (៤-៦-៧ ព្យាង្គ)",
        "energy": "មហិមា វិសេសវិសាល",
        "dna": "Ancient Angkor Victory & Bronze Gong Harmonic Purity",
        "inst": ["Thunderous Skor Thom", "Piercing Sralai Oboe", "War Brass"],
        "nouns": ["ព្រះខាន់មាស", "អាវក្រោះដែក", "ទង់ជ័យមហានគរ", "កំពែងថ្មបុរាណ"],
        "modifiers": ["ដ៏អង់អាចក្លាហាន", "ភ្លឺចែងចាំងកណ្តាលសឹក", "មិនចេះរាថយ", "នៃដែនដីសុវណ្ណភូមិ"]
    },
    "កតញ្ញូតាធម៌ដ៏ជ្រាលជ្រៅ": {
        "genre": "[24-bit 48kHz] Traditional Khmer acoustic ballad, Khim dulcimer, warm guitar, 72 BPM",
        "vocal": "ប្រុស (មាំទាំ រណ្តំ)",
        "chhanda": "បទពាក្យ ៨ (៤+៤ ព្យាង្គ)",
        "energy": "កម្សត់ស្រទន់ រណ្តំចិត្ត",
        "dna": "Filial Piety, Mother's Grace, and Golden Thread Memories",
        "inst": ["Sweet Khim Dulcimer", "Warm Acoustic Guitar", "Soft Khloy"],
        "nouns": ["ស្បៃសូត្រមាតា", "ដំណក់ញើសបិតា", "ដំបូលផ្ទះចាស់", "ចង្កៀងប្រេងកាត"],
        "modifiers": ["ដ៏កក់ក្តៅបំផុត", "ធំធេងគ្មានព្រំដែន", "ដែលចិញ្ចឹមបីបាច់", "ពេញដោយក្តីមេត្តា"]
    },
    "ស្នេហាផ្អែមល្ហែមសម័យមាស": {
        "genre": "[24-bit 48kHz] 1960s Cambodian rock, twangy surf guitar, retro Farfisa organ, 108 BPM",
        "vocal": "ឆ្លើយឆ្លង (ប្រុស & ស្រី)",
        "chhanda": "បទពាក្យ ៧ (៧ ព្យាង្គ)",
        "energy": "រ៉ូមែនទិក ផ្អែមល្ហែម",
        "dna": "1960s Battambang Sangker River Romance & Vespa Ride",
        "inst": ["Twangy Surf Guitar", "Retro Farfisa Organ", "Walking Bassline"],
        "nouns": ["មាត់ស្ទឹងសង្កែ", "ម៉ូតូវ៉េស្ប៉ាចាស់", "សំបុត្រស្នេហ៍", "ផ្កាចម្ប៉ា"],
        "modifiers": ["នាទសវត្សរ៍ទី ៦០", "ក្រអូបឈ្ងុយឈ្ងប់", "ដ៏ផ្អែមល្ហែមត្រជាក់ចិត្ត", "ពេលថ្ងៃរៀបលិច"]
    },
    "អាថ៌កំបាំងព្រឺព្រួច": {
        "genre": "[24-bit 48kHz] Dark cinematic synthwave, analog arpeggiators, rain on zinc, 90 BPM",
        "vocal": "និទានរឿង (បែបភាពយន្ត)",
        "chhanda": "បទពាក្យ ៩ / កម្រងកែវ",
        "energy": "ស្វាហាប់ រស់រវើក",
        "dna": "Midnight Temple Shadows & Ancient Stone Inscriptions",
        "inst": ["Dark Analog Synth", "Haunting Bowed Chapei", "Rain on Zinc Roof"],
        "nouns": ["ស្រមោលអតីតកាល", "សិលាចារឹកអាថ៌កំបាំង", "អ័ព្ទរាត្រីស្ងាត់", "ស្នូរជើងកណ្តាលយប់"],
        "modifiers": ["ដ៏ព្រឺព្រួចត្រជាក់ស្រេប", "លាក់ទុករាប់សតវត្សរ៍", "ក្រោមពន្លឺចន្ទគ្រាស", "ដែលគ្មានអ្នកដឹង"]
    }
}

POPULAR_STORY_KEYWORDS = [
    "អង្គរវត្ត, រង់ចាំស្នេហ៍ស្មោះ, ទឹកភ្នែកកណ្តាលថ្មភក់",
    "ស្ទឹងសង្កែ, ម៉ូតូវ៉េស្ប៉ា, សំបុត្រស្នេហ៍ទសវត្សរ៍ ៦០",
    "រាត្រីភ្នំពេញ, ភ្លើងនីអុង, បាសបុក 808 Trap",
    "ការបែកបាក់, ញញឹមទាំងទឹកភ្នែក, សន្យាព្រហ្មលិខិត",
    "មេទ័ពសឹក, អាវក្រោះដែក, ផ្កាចម្ប៉ាអមផ្លូវចម្បាំង",
    "ភ្លៀងធ្លាក់លើដំបូលស័ង្កសី, កង្ហារចាស់, នឹកអ្នកនៅឆ្ងាយ"
]

INTENT_MODES = [
    "Mode 1: ផលិតបទថ្មីពេញលេញ",
    "Mode 2: គោះព្យាង្គ ១ទល់១ ច្បាស់ៗ",
    "Mode 3: កែសម្រួលដោយរក្សាចង្វាក់ជួន",
    "Mode 4: ប្តូរតួអង្គ (បង-អូន / សំឡេង)",
    "Mode 5: បំប្លែងជាភ្លេងភាពយន្ត (OST)"
]

KHMER_PHONETIC_DICTIONARY = {
    "កិត្តិយស": "កិត-តិ-យស់",
    "មហានគរ": "មៈ-ហា-នៈ-គរ",
    "អាណាចក្រ": "អា-ណា-ចាក់",
    "ប្រវត្តិសាស្ត្រ": "ប្រៈ-វ៉ាត់-តិ-សាស",
    "ព្រហ្មលិខិត": "ព្រំ-លិ-ខិត",
    "សច្ចា": "សាច់-ចា",
    "អរុណរះ": "អៈ-រ៉ុន-រះ",
    "កតញ្ញូ": "កៈ-តាញ់-ញូ"
}

TIME_PHRASES = [
    "នាពេលរាត្រីស្ងាត់", "ក្រោមពន្លឺព្រះចន្ទពេញបូណ៌មី", "នាព្រឹកព្រលឹមស្រស់បំព្រង",
    "ពេលថ្ងៃរៀបលិចលើជើងមេឃ", "ក្នុងរដូវផ្ការីកក្រអូបឈ្ងុយ", "កណ្តាលជំនោរខ្យល់ត្រជាក់", "ក្បែរកំពែងថ្មបុរាណ"
]
EMOTION_PHRASES = [
    "ដួងចិត្តរំភើបញាប់ញ័រឥតឧបមា", "បេះដូងភក្ដីចារឹកសច្ចាមិនប្រែប្រួល", "វិញ្ញាណស្នេហ៍ហោះហើរឆ្លងកាត់ពេលវេលា",
    "ក្ដីសង្ឃឹមភ្លឺចែងចាំងដូចពេជ្រចរណៃ", "មនោសញ្ចេតនាជ្រាលជ្រៅដក់ជាប់ក្នុងទ្រូង", "សូរសៀងដូនតាប្រទានពរជ័យមង្គល"
]
ACTION_PHRASES = [
    "ប្រគំបទភ្លេងថ្មីកំដរដួងព្រលឹង", "ឆ្លាក់ទុកជាស្នាដៃកម្មសិទ្ធិបញ្ញាដ៏ឧត្តុង្គឧត្តម", "បំភ្លឺផ្លូវជីវិតឆ្ពោះទៅកាន់ត្រើយសុភមង្គល",
    "ថែរក្សាមរតកវប្បធម៌ឱ្យគង់វង្សនិរន្តរ៍", "បង្កើតនិស្ស័យស្នេហ៍បរិសុទ្ធគ្មានថ្ងៃរលត់", "លាន់ឮខ្ទរខ្ទារពេញដែនដីសុវណ្ណភូមិ"
]

def apply_phonetic_bypass(text: str) -> str:
    for raw_word, phonetic_word in KHMER_PHONETIC_DICTIONARY.items():
        text = text.replace(raw_word, phonetic_word)
    return text

def audit_lyric_engine_bars(lyrics_text: str, chhanda_form: str = "") -> str:
    lines = lyrics_text.split('\n')
    audited = []
    for line in lines:
        stripped = line.strip()
        if stripped.startswith('[') or stripped.startswith('(') or not stripped:
            audited.append(line)
            continue
        audited.append(apply_phonetic_bypass(line))
    return "\n".join(audited)

def make_unique_khmer_line(base_prefix: str, nouns: list, modifiers: list) -> str:
    for _ in range(100):
        t_p = random.choice(TIME_PHRASES)
        e_p = random.choice(EMOTION_PHRASES)
        a_p = random.choice(ACTION_PHRASES)
        n_p = random.choice(nouns)
        m_p = random.choice(modifiers)
        candidate_ln = f"{base_prefix} {n_p}{m_p} | {t_p} {e_p} | {a_p}។"
        h = compute_sha256_hex(candidate_ln)
        if h not in LOCKED_LINE_HASHES:
            LOCKED_LINE_HASHES.add(h)
            return candidate_ln
    return f"{base_prefix} សូរសៀងថ្មីផ្តាច់មុខបំភ្លឺផ្លូវអារម្មណ៍ #{random.randint(1000, 9999)}។"

def run_custom_local_lyric_engine(theme, genre_style, mood, vocal_profile, chhanda, energy, dna_memory, thinking_depth):
    cfg = MOOD_LOGIC_MATRIX.get(mood, list(MOOD_LOGIC_MATRIX.values())[0])
    inst_triplet = cfg["inst"]
    nouns = cfg["nouns"]
    modifiers = cfg["modifiers"]

    is_duet = "ឆ្លើយឆ្លង" in vocal_profile or "duet" in vocal_profile.lower()
    is_rap = "រ៉េប" in vocal_profile or "rap" in vocal_profile.lower()
    is_narr = "និទាន" in vocal_profile or "narration" in vocal_profile.lower()

    v_cue_1 = "[Verse 1: Duet Call-and-Response]" if is_duet else ("[Verse 1: Khmer Rap Flow]" if is_rap else ("[Verse 1: Cinematic Narration]" if is_narr else "[Verse 1: Emotive Vocal]"))
    v_cue_2 = "[Verse 2: Duet Counterpoint]" if is_duet else ("[Verse 2: Rap Cadence]" if is_rap else "[Verse 2: Melodic Phrasing]")
    ch_cue = "[Chorus: Anthem Harmony]" if is_duet else "[Chorus: Dynamic Crescendo]"

    story_clean = theme.strip() if theme.strip() else "សច្ចាស្នេហ៍មហានគរ"
    track_id = random.randint(1000, 9999)
    cand_title = f"សច្ចាស្នេហ៍ {random.choice(nouns)} | ({story_clean[:22]} #{track_id})"
    b1_final = enforce_box1_limit(genre_style, 120)

    stanza_blocks = [
        build_diction_header(vocal_profile),
        "",
        f"[Intro: {inst_triplet[0]} & {inst_triplet[1]} Opening]",
        f"(ស្នាដៃនិពន្ធថ្មី — {story_clean[:45]} | កម្រិត៖ {thinking_depth} | {chhanda})",
        "",
        v_cue_1,
        f"ខ្យល់បក់រំភើយ កាត់តាមជំនោរ {story_clean}",
        make_unique_khmer_line("ក្រោមមេឃាដ៏ល្វឹងល្វើយនៃ", nouns, modifiers),
        make_unique_khmer_line("សំឡេងតន្ត្រីរំលេចនូវ", nouns, modifiers),
        "",
        v_cue_2,
        make_unique_khmer_line("គ្រប់ជំហាននៃជីវិតឆ្លងកាត់", nouns, modifiers),
        make_unique_khmer_line("ពន្លឺអរុណរះបំភ្លឺ", nouns, modifiers),
        ""
    ]

    stanza_blocks.extend([
        ch_cue,
        "មហានគរអាណាចក្រស្នេហ៍ដ៏ស្មោះស្ម័គ្រ!",
        "សូរសៀងដង្ហោយហៅ ឆ្លងកាត់ភពផែនដី",
        make_unique_khmer_line("យើងរួមគ្នាថែរក្សា កិត្តិយស និង", nouns, modifiers),
        "ទោះផែនដីវិលផ្លាស់ ក៏ដួងចិត្តមិនប្រែប្រួលឡើយ!",
        ""
    ])

    stanza_blocks.extend([
        f"[Outro: {inst_triplet[-1]} Final Breath]",
        "ស្នាមញញឹមចុងក្រោយ រលាយក្នុងអ័ព្ទរាត្រី...",
        "[Fade out]",
        "[End]"
    ])

    box2_final = audit_lyric_engine_bars("\n".join(stanza_blocks), chhanda)
    return cand_title, b1_final, box2_final

def vo5_updater_and_refine_logic(
    current_title, current_box1, current_box2, refinement_instruction, vocal_str, intent_mode,
    theme, mood, chhanda, energy, dna_memory, thinking_depth
):
    if not current_box2 or not current_box2.strip() or "Mode 1" in intent_mode:
        t, b1, b2 = generate_track_ui(theme, current_box1, mood, vocal_str, chhanda, energy, dna_memory, thinking_depth)
        return t, b1, b2, f"[Mode 1] ផលិតបទថ្មីជោគជ័យ! (ទម្រង់ភ្លេង៖ {len(b1)}/120 តួអក្សរ)"

    updated_title = current_title.strip() if current_title else "សច្ចាស្នេហ៍មហានគរ (កែសម្រួលថ្មី)"
    updated_b1 = current_box1.strip()
    updated_b2 = current_box2.strip()
    instr_clean = refinement_instruction.strip()

    if "Mode 2" in intent_mode:
        updated_b2 = apply_phonetic_bypass(updated_b2)
        updated_b2 = re.sub(r"\s+\|\s+", " • ", updated_b2)
        status_msg = "[Mode 2] គោះព្យាង្គ ១ទល់១ រួចរាល់!"
    elif "Mode 4" in intent_mode:
        swap_map = {"បង": "អូន", "អូន": "បង", "ប្រុស": "ស្រី", "ស្រី": "ប្រុស"}
        for k, v in swap_map.items():
            updated_b2 = updated_b2.replace(k, v)
        status_msg = "[Mode 4] ប្តូរតួអង្គ (បង-អូន) រួចរាល់!"
    elif "Mode 5" in intent_mode:
        updated_b1 = enforce_box1_limit(f"{updated_b1}, orchestral OST", 120)
        updated_b2 += "\n\n[Orchestral Climax: Symphony]\n[Fade out]\n[End]"
        status_msg = "[Mode 5] បំប្លែងជាភ្លេងភាពយន្ត (OST) រួចរាល់!"
    else:
        if instr_clean:
            updated_b2 += f"\n\n({instr_clean})\n[Fade out]\n[End]"
        status_msg = "[Mode 3] កែសម្រួលពាក្យពេចន៍រួចរាល់!"

    updated_b1 = enforce_box1_limit(updated_b1, 120)
    return updated_title, updated_b1, updated_b2, f"{status_msg} (ទម្រង់ភ្លេង៖ {len(updated_b1)}/120 តួអក្សរ)"

def on_mood_change(selected_mood):
    config = MOOD_LOGIC_MATRIX.get(selected_mood, list(MOOD_LOGIC_MATRIX.values())[0])
    return config["genre"], config["vocal"], config["chhanda"], config["energy"], config["dna"]

def update_story_suggestions(query):
    if not query:
        return gr.update(choices=POPULAR_STORY_KEYWORDS, visible=True)
    filtered = [item for item in POPULAR_STORY_KEYWORDS if query.lower() in item.lower()]
    if not filtered:
        filtered = [f"បង្កើតថ្មី៖ {query}", query]
    return gr.update(choices=filtered, visible=True)

def select_story_suggestion(evt: gr.SelectData):
    return evt.value.replace("បង្កើតថ្មី៖ ", "")

def enforce_box1_limit(style_str: str, max_len: int = 120) -> str:
    s = str(style_str).strip().replace("\n", " ")
    if not s.startswith("[24-bit 48kHz]"):
        s = "[24-bit 48kHz] " + re.sub(r"^\[.*?\]\s*", "", s)
    if len(s) <= max_len:
        return s
    truncated = s[:max_len]
    if "," in truncated:
        truncated = truncated.rsplit(",", 1)[0]
    elif " " in truncated:
        truncated = truncated.rsplit(" ", 1)[0]
    return truncated.strip()

def parse_and_validate_vo5_output(raw_text: str, chhanda_form: str = "", vocal_str: str = "", **kwargs) -> dict:
    title_match = re.search(r"===\s*(?:SONG\s*)?TITLE\s*===\s*\n+(.*?)(?=\n+===|\Z)", raw_text, re.S | re.I)
    box1_match = re.search(r"===\s*BOX 1.*?===\s*\n+(.*?)(?=\n+===|\Z)", raw_text, re.S | re.I)
    box2_match = re.search(r"===\s*BOX 2.*?===\s*\n+(.*)", raw_text, re.S | re.I)

    title = title_match.group(1).strip() if title_match else "សច្ចាស្នេហ៍មហានគរ"
    raw_box1 = box1_match.group(1).strip() if box1_match else "[24-bit 48kHz] Epic Angkor cinematic ballad, 70 BPM"
    raw_box2 = box2_match.group(1).strip() if box2_match else raw_text.strip()

    valid_box1 = enforce_box1_limit(raw_box1, 120)
    header_str = build_diction_header(vocal_str)

    if "[Vocal Mix: 24-bit" not in raw_box2:
        valid_box2 = f"{header_str}\n\n{raw_box2}"
    else:
        valid_box2 = raw_box2

    valid_box2 = audit_lyric_engine_bars(valid_box2, chhanda_form)

    return {
        "title": title,
        "box1": valid_box1,
        "box1_len": len(valid_box1),
        "box2": valid_box2
    }

def generate_track_ui(theme, genre_style, mood, vocal_profile, chhanda, energy, dna_memory, thinking_depth):
    prompt = f"""[VO5 STUDIO]
- សាច់រឿង: {theme}
- ប្រភេទភ្លេង: {genre_style}
- អារម្មណ៍: {mood}
- សំឡេង: {vocal_profile}
- រង្វាស់ព្យាង្គ: {chhanda}
Format output with:
=== TITLE ===
<Song Title>
=== BOX 1 ===
<Style <= 120 chars>
=== BOX 2 ===
<Lyrics>"""
    try:
        response = client.models.generate_content(
            model=MODEL_ID,
            contents=prompt,
            config=types.GenerateContentConfig(
                system_instruction=MASTER_KERNEL_10_0,
                thinking_config=types.ThinkingConfig(thinking_level=types.ThinkingLevel.HIGH)
            )
        )
        res = parse_and_validate_vo5_output(response.text, chhanda_form=chhanda, vocal_str=vocal_profile)
        return res["title"], res["box1"], res["box2"]
    except Exception:
        return run_custom_local_lyric_engine(theme, genre_style, mood, vocal_profile, chhanda, energy, dna_memory, thinking_depth)

def generate_track_with_status(theme, genre_style, mood, vocal_profile, chhanda, energy, dna_memory, thinking_depth):
    t, b1, b2 = generate_track_ui(theme, genre_style, mood, vocal_profile, chhanda, energy, dna_memory, thinking_depth)
    status = f"ជោគជ័យ! | ទម្រង់ភ្លេង៖ {len(b1)}/120 តួអក្សរ"
    return t, b1, b2, status

# ==============================================================================
# 6. iOS HIG-COMPLIANT CSS & RESPONSIVE UI (DARK MODE & TOUCH TARGETS)
# ==============================================================================
ios_dark_css = """
:root {
    color-scheme: dark;
}
.gradio-container {
    background-color: #000000 !important;
    color: #f5f5f7 !important;
    font-family: -apple-system, BlinkMacSystemFont, "SF Pro Text", "SF Pro Icons", "Helvetica Neue", Helvetica, Arial, sans-serif !important;
    max-width: 430px !important;
    margin: 0 auto !important;
    padding: 12px !important;
}

.studio-card {
    background: rgba(28, 28, 30, 0.75) !important;
    backdrop-filter: blur(20px) !important;
    -webkit-backdrop-filter: blur(20px) !important;
    border: 0.5px solid rgba(255, 255, 255, 0.15) !important;
    border-radius: 16px !important;
    padding: 16px !important;
    margin-bottom: 12px !important;
}

h2, h3, .gr-markdown h3 {
    font-size: 17px !important;
    font-weight: 600 !important;
    letter-spacing: -0.41px !important;
    color: #ffffff !important;
    margin-bottom: 10px !important;
}

label span, .block label {
    font-size: 13px !important;
    font-weight: 500 !important;
    color: #8e8e93 !important;
    margin-bottom: 4px !important;
}

input, textarea, select, .gr-box, .gr-input {
    background-color: #2c2c2e !important;
    border: 0.5px solid #3a3a3c !important;
    border-radius: 10px !important;
    color: #ffffff !important;
    font-size: 15px !important;
    min-height: 44px !important;
    padding: 10px 12px !important;
}

input:focus, textarea:focus {
    border-color: #0a84ff !important;
    box-shadow: 0 0 0 2px rgba(10, 132, 255, 0.3) !important;
}

#gen-btn {
    background: linear-gradient(135deg, #f59e0b 0%, #d97706 100%) !important;
    color: #000000 !important;
    font-weight: 600 !important;
    font-size: 16px !important;
    border-radius: 12px !important;
    height: 50px !important;
    border: none !important;
    box-shadow: 0 4px 12px rgba(245, 158, 11, 0.3) !important;
    cursor: pointer;
}

#refine-btn {
    background: #0a84ff !important;
    color: #ffffff !important;
    font-weight: 600 !important;
    border-radius: 10px !important;
    height: 44px !important;
    border: none !important;
    cursor: pointer;
}

.gr-code {
    background-color: #1c1c1e !important;
    border-radius: 10px !important;
    border: 0.5px solid #3a3a3c !important;
}
"""

with gr.Blocks(title="VO5 Studio iOS", css=ios_dark_css) as demo:
    gr.HTML("""
    <div style="background: rgba(44, 44, 46, 0.6); backdrop-filter: blur(10px); border: 0.5px solid rgba(255, 255, 255, 0.1); padding: 14px; border-radius: 14px; margin-bottom: 14px; text-align: center;">
        <span style="background: #3a3a3c; color: #f59e0b; padding: 2px 10px; border-radius: 8px; font-size: 11px; font-weight: 600; text-transform: uppercase;">iOS Optimized UI</span>
        <h2 style="margin: 6px 0 2px 0; color: #ffffff; font-size: 20px; font-weight: 700;">VO5 ប្រព័ន្ធនិពន្ធចម្រៀង</h2>
        <p style="margin: 0; color: #8e8e93; font-size: 12px;">កំណែ 10.0.0 | រចនាបទទំនើបសម្រាប់ទូរស័ព្ទ</p>
    </div>
    """)

    with gr.Column(elem_classes=["studio-card"]):
        gr.Markdown("### កំណត់រចនាសម្ព័ន្ធបទចម្រៀង")

        theme_input = gr.Textbox(
            label="សាច់រឿង ឬអត្ថន័យបទ",
            placeholder="បញ្ចូលអត្ថន័យសង្ខេប...",
            lines=2,
            value="សេចក្តីស្រឡាញ់ដ៏ស្មោះស្ម័គ្រ និងការរង់ចាំនៅមុខប្រាសាទអង្គរវត្ត"
        )

        story_dropdown = gr.Dropdown(
            label="ឃ្លាប្រយោគពេញនិយម",
            choices=POPULAR_STORY_KEYWORDS,
            interactive=True
        )

        mood_dropdown = gr.Dropdown(
            label="អារម្មណ៍បទភ្លេង",
            choices=list(MOOD_LOGIC_MATRIX.keys()),
            value="ឈឺចាប់ស៊ីជម្រៅ"
        )

        genre_input = gr.Textbox(
            label="ប្រភេទភ្លេង និងឧបករណ៍",
            value=MOOD_LOGIC_MATRIX["ឈឺចាប់ស៊ីជម្រៅ"]["genre"]
        )

        vocal_dropdown = gr.Dropdown(
            label="សំឡេងអ្នកចម្រៀង",
            choices=[
                "ស្រី (ផ្អែមស្រទន់)",
                "ប្រុស (មាំទាំ រណ្តំ)",
                "ឆ្លើយឆ្លង (ប្រុស & ស្រី)",
                "រ៉េប (Khmer Rap)",
                "និទានរឿង (បែបភាពយន្ត)"
            ],
            value=MOOD_LOGIC_MATRIX["ឈឺចាប់ស៊ីជម្រៅ"]["vocal"]
        )

        chhanda_dropdown = gr.Dropdown(
            label="រង្វាស់ព្យាង្គ (ច្បាប់កាព្យ)",
            choices=[
                "បទពាក្យ ៧ (៧ ព្យាង្គ)",
                "បទកាកគតិ (៤ ព្យាង្គ)",
                "បទព្រហ្មគីតិ (៥-៦ ព្យាង្គ)",
                "បទភុជង្គលីលា (៤-៦-៧ ព្យាង្គ)",
                "បទពាក្យ ៨ (៤+៤ ព្យាង្គ)",
                "បទពាក្យ ៩ / កម្រងកែវ"
            ],
            value=MOOD_LOGIC_MATRIX["ឈឺចាប់ស៊ីជម្រៅ"]["chhanda"]
        )

        energy_dropdown = gr.Dropdown(
            label="កម្រិតថាមពល",
            choices=[
                "មហិមា វិសេសវិសាល",
                "កម្សត់ស្រទន់ រណ្តំចិត្ត",
                "ស្វាហាប់ រស់រវើក",
                "រ៉ូមែនទិក ផ្អែមល្ហែម"
            ],
            value=MOOD_LOGIC_MATRIX["ឈឺចាប់ស៊ីជម្រៅ"]["energy"]
        )

        dna_memory_input = gr.Textbox(
            label="អត្តសញ្ញាណភ្លេង (DNA)",
            value=MOOD_LOGIC_MATRIX["ឈឺចាប់ស៊ីជម្រៅ"]["dna"]
        )

        thinking_radio = gr.Radio(
            label="កម្រិតគិតរបស់ AI",
            choices=["LOW", "MEDIUM", "HIGH"],
            value="HIGH"
        )

        generate_btn = gr.Button("បង្កើតបទចម្រៀង", elem_id="gen-btn")

    with gr.Column(elem_classes=["studio-card"]):
        gr.Markdown("### លទ្ធផលទំនុកច្រៀង និងទម្រង់ភ្លេង")
        status_banner = gr.Markdown("ចុច **បង្កើតបទចម្រៀង** ដើម្បីចាប់ផ្តើម")
        title_box = gr.Code(label="១. ចំណងជើងបទចម្រៀង", lines=2)
        box1_box = gr.Code(label="២. ទម្រង់តន្ត្រី (យ៉ាងច្រើន ១២០ តួអក្សរ)", lines=2)
        box2_box = gr.Code(label="៣. ទំនុកច្រៀងពេញ និងការបែងចែកព្យាង្គ", lines=10)

        gr.Markdown("---")
        gr.Markdown("### ផ្ទាំងកែសម្រួល និងអាប់ដេត")
        refine_mode_dropdown = gr.Dropdown(
            label="របៀបកែសម្រួល",
            choices=INTENT_MODES,
            value=INTENT_MODES[2]
        )
        refine_input = gr.Textbox(
            label="បញ្ជាកែលម្អបន្ថែម",
            value="កែសម្រួលពាក្យពេចន៍ឱ្យកាន់តែស៊ីជម្រៅ និងរក្សាទម្រង់ភ្លេង <= ១២០ តួអក្សរ",
            lines=2
        )
        refine_btn = gr.Button("កែសម្រួល និងអាប់ដេត", elem_id="refine-btn")

    # Events
    theme_input.input(fn=update_story_suggestions, inputs=[theme_input], outputs=[story_dropdown])
    story_dropdown.select(fn=select_story_suggestion, outputs=[theme_input])
    mood_dropdown.change(fn=on_mood_change, inputs=[mood_dropdown], outputs=[genre_input, vocal_dropdown, chhanda_dropdown, energy_dropdown, dna_memory_input])

    generate_btn.click(
        fn=generate_track_with_status,
        inputs=[theme_input, genre_input, mood_dropdown, vocal_dropdown, chhanda_dropdown, energy_dropdown, dna_memory_input, thinking_radio],
        outputs=[title_box, box1_box, box2_box, status_banner]
    )

    refine_btn.click(
        fn=vo5_updater_and_refine_logic,
        inputs=[title_box, box1_box, box2_box, refine_input, vocal_dropdown, refine_mode_dropdown, theme_input, mood_dropdown, chhanda_dropdown, energy_dropdown, dna_memory_input, thinking_radio],
        outputs=[title_box, box1_box, box2_box, status_banner]
    )

demo.launch(share=True, debug=False)
````
